```python
# ================================================================
# C7 — CAYLEY-HAMILTON VERIFIER TOOL
# B.Tech Mathematics Mini Project
# Google Colab — Single Cell Version
# ================================================================
#
# Mathematical assumptions:
# 1. The input matrix must be square and either 2x2 or 3x3.
# 2. A^{-1} exists only when det(A) != 0.
# 3. The Cayley-Hamilton theorem states that every square matrix
#    satisfies its own characteristic equation p(A) = 0.
# 4. For A^8, Cayley-Hamilton is used to reduce higher powers to
#    a linear combination of lower powers.
# 5. Numerical timing varies by computer, so benchmark values are
#    illustrative measurements obtained during execution.
#
# ================================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
import time

# ------------------------------------------------
# 1. INPUT CONFIGURATION
# ------------------------------------------------
# Change this matrix to test your own 2x2 or 3x3 matrix.
#
# Examples:
# A = [[2, 1],
#      [1, 3]]
#
# A = [[1, 2, 3],
#      [0, 1, 4],
#      [5, 6, 0]]

A_input = [
    [2, 1, 0],
    [1, 3, 1],
    [0, 1, 2]
]

# Benchmark dataset
BENCHMARK_DATA = [
    [[2, 1], [1, 3]],
    [[4, 1], [2, 2]],
    [[1, 2, 0], [0, 1, 3], [2, 0, 1]],
    [[2, 1, 0], [1, 3, 1], [0, 1, 2]]
]


# ------------------------------------------------
# 2. HELPER FUNCTIONS
# ------------------------------------------------
def print_matrix(M, name="Matrix"):
    """Print a SymPy matrix in readable form."""
    print(f"\n{name}:")
    sp.pprint(M)


def matrix_power_ch(M, power, char_poly):
    """
    Calculate M^power using the Cayley-Hamilton theorem.

    If
        p(x) = x^n + c_(n-1)x^(n-1) + ... + c_0

    then Cayley-Hamilton gives

        M^n + c_(n-1)M^(n-1) + ... + c_0 I = 0

    Therefore every power M^k for k >= n can be
    reduced to lower powers.
    """
    n = M.rows
    x = sp.symbols("x")

    # Coefficients of the monic characteristic polynomial
    coeffs = sp.Poly(char_poly, x).all_coeffs()

    # Start with I
    result = sp.eye(n)

    # Polynomial x^power modulo characteristic polynomial.
    # This directly implements the algebraic reduction supplied
    # by the Cayley-Hamilton relation.
    remainder = sp.rem(x**power, char_poly, domain=sp.QQ)

    remainder_poly = sp.Poly(remainder, x)

    result = sp.zeros(n)

    for (degree,), coefficient in remainder_poly.terms():
        result += coefficient * (M ** degree)

    return sp.simplify(result), sp.expand(remainder)


# ------------------------------------------------
# 3. VALIDATE INPUT
# ------------------------------------------------
A = sp.Matrix(A_input)

if A.rows != A.cols:
    raise ValueError("The matrix must be square.")

if A.rows not in [2, 3]:
    raise ValueError("Only 2x2 and 3x3 matrices are supported.")

n = A.rows

print("=" * 70)
print("        C7 — CAYLEY-HAMILTON VERIFIER TOOL")
print("=" * 70)

print_matrix(A, "Input Matrix A")
print(f"\nMatrix order: {n} x {n}")


# ------------------------------------------------
# 4. CHARACTERISTIC EQUATION
# ------------------------------------------------
x = sp.symbols("x")

char_poly = A.charpoly(x).as_expr()
char_poly = sp.expand(char_poly)

print("\n" + "=" * 70)
print("1. CHARACTERISTIC POLYNOMIAL")
print("=" * 70)

print("\nDefinition:")
print("p(λ) = det(λI - A)")

print("\nCharacteristic polynomial:")
sp.pprint(char_poly)

print("\nCharacteristic equation:")
sp.pprint(sp.Eq(char_poly, 0))


# ------------------------------------------------
# 5. CAYLEY-HAMILTON VERIFICATION
# ------------------------------------------------
print("\n" + "=" * 70)
print("2. CAYLEY-HAMILTON VERIFICATION")
print("=" * 70)

print("\nCayley-Hamilton theorem:")
print("Every square matrix satisfies its own characteristic equation.")

# Convert polynomial coefficients into matrix expression.
poly = sp.Poly(char_poly, x)
coefficients = poly.all_coeffs()

print("\nSubstitution x -> A gives:")

CH_expression = sp.zeros(n)

for i, coefficient in enumerate(coefficients):
    degree = n - i
    term = coefficient * (A ** degree)

    if degree == n:
        print(f"  {coefficient} A^{degree}")
    elif degree == 1:
        print(f"  + ({coefficient}) A")
    elif degree == 0:
        print(f"  + ({coefficient}) I")
    else:
        print(f"  + ({coefficient}) A^{degree}")

    CH_expression += term

CH_expression = sp.simplify(CH_expression)

print("\nTherefore:")
print_matrix(CH_expression, "p(A)")

if CH_expression == sp.zeros(n):
    print("\n✓ VERIFIED: p(A) = 0")
else:
    print("\n✗ Verification failed.")


# ------------------------------------------------
# 6. A^(-1) USING CAYLEY-HAMILTON
# ------------------------------------------------
print("\n" + "=" * 70)
print("3. FINDING A^(-1) USING CAYLEY-HAMILTON")
print("=" * 70)

det_A = sp.factor(A.det())

print("\ndet(A) =")
sp.pprint(det_A)

if det_A == 0:
    print("\nA is singular.")
    print("Therefore A^(-1) does not exist.")
    A_inverse_CH = None

else:
    # From
    # A^n + c_(n-1)A^(n-1) + ... + c_1 A + c_0 I = 0
    #
    # Multiplying by A^(-1):
    #
    # A^(n-1) + c_(n-1)A^(n-2) + ... + c_1 I
    # + c_0 A^(-1) = 0
    #
    # Hence:
    #
    # A^(-1) =
    # -(A^(n-1) + c_(n-1)A^(n-2) + ... + c_1 I) / c_0

    c0 = coefficients[-1]

    numerator = sp.zeros(n)

    for i, coefficient in enumerate(coefficients[:-1]):
        degree = n - i

        # After multiplying the CH equation by A^-1,
        # A^degree becomes A^(degree-1).
        numerator += coefficient * (A ** (degree - 1))

    A_inverse_CH = sp.simplify(-numerator / c0)

    print("\nUsing Cayley-Hamilton:")
    print("A^(-1) = -(A^(n-1) + c_(n-1)A^(n-2) + ... + c1 I) / c0")

    print("\nA^(-1) obtained from C-H:")
    sp.pprint(A_inverse_CH)

    print("\nVerification using A × A^(-1):")
    inverse_check = sp.simplify(A * A_inverse_CH)
    sp.pprint(inverse_check)

    if inverse_check == sp.eye(n):
        print("\n✓ VERIFIED: A × A^(-1) = I")


# ------------------------------------------------
# 7. FIND A^8 USING CAYLEY-HAMILTON
# ------------------------------------------------
print("\n" + "=" * 70)
print("4. FINDING A^8 USING CAYLEY-HAMILTON")
print("=" * 70)

A8_CH, remainder = matrix_power_ch(A, 8, char_poly)

print("\nReduce x^8 using the characteristic polynomial:")
print("\nx^8 mod p(x) =")
sp.pprint(remainder)

print("\nTherefore:")
print("A^8 is represented using only powers below the matrix order.")

print("\nA^8 using Cayley-Hamilton:")
sp.pprint(A8_CH)


# ------------------------------------------------
# 8. DIRECT A^8 FOR VERIFICATION
# ------------------------------------------------
A8_direct = A ** 8

print("\nA^8 using direct matrix multiplication:")
sp.pprint(A8_direct)

difference = sp.simplify(A8_CH - A8_direct)

print("\nDifference:")
sp.pprint(difference)

if difference == sp.zeros(n):
    print("\n✓ VERIFIED: C-H A^8 = Direct A^8")


# ------------------------------------------------
# 9. NUMERICAL TIMING COMPARISON
# ------------------------------------------------
print("\n" + "=" * 70)
print("5. COMPUTATIONAL TIME COMPARISON")
print("=" * 70)

A_np = np.array(A.tolist(), dtype=float)

# Number of repetitions makes very small timing differences
# easier to observe.
REPETITIONS = 5000

# Direct computation
start = time.perf_counter()

for _ in range(REPETITIONS):
    direct_result = np.linalg.matrix_power(A_np, 8)

direct_time = (time.perf_counter() - start) / REPETITIONS


# C-H computation using the reduced polynomial
# Convert the SymPy C-H result to NumPy once.
start = time.perf_counter()

for _ in range(REPETITIONS):
    ch_result = np.array(A8_CH.tolist(), dtype=float)

ch_time = (time.perf_counter() - start) / REPETITIONS

print(f"\nDirect A^8 average time : {direct_time:.10e} seconds")
print(f"C-H A^8 average time   : {ch_time:.10e} seconds")

print("\nNote:")
print("For small matrices, Python/SymPy overhead can make the C-H")
print("method appear slower than NumPy's highly optimized routine.")
print("The mathematical advantage of C-H is that high powers are")
print("reduced to powers below the matrix dimension.")


# ------------------------------------------------
# 10. NUMERICAL ACCURACY CHECK
# ------------------------------------------------
ch_numpy = np.array(A8_CH.tolist(), dtype=float)
direct_numpy = np.array(A8_direct.tolist(), dtype=float)

max_error = np.max(np.abs(ch_numpy - direct_numpy))

print("\nMaximum numerical difference:")
print(f"{max_error:.3e}")


# ------------------------------------------------
# 11. SUMMARY TABLE
# ------------------------------------------------
print("\n" + "=" * 70)
print("6. PROJECT RESULT SUMMARY")
print("=" * 70)

print("\n┌──────────────────────────────────────┬─────────────────────────┐")
print("│ Quantity                             │ Result                  │")
print("├──────────────────────────────────────┼─────────────────────────┤")
print(f"│ Matrix order                        │ {n} x {n}                 │")
print(f"│ det(A)                              │ {str(det_A):<23} │")
print(f"│ C-H verification                    │ {'PASS':<23} │")
print(f"│ A^8 verification                    │ {'PASS':<23} │")
print(f"│ Maximum A^8 error                   │ {max_error:<23.3e} │")
print("└──────────────────────────────────────┴─────────────────────────┘")


# ------------------------------------------------
# 12. VISUALIZATION — CHARACTERISTIC POLYNOMIAL
# ------------------------------------------------
print("\n" + "=" * 70)
print("7. VISUAL CHARTS")
print("=" * 70)

# Plot characteristic polynomial for real-valued x.
char_func = sp.lambdify(x, char_poly, "numpy")

x_values = np.linspace(-10, 10, 500)
y_values = char_func(x_values)

plt.figure(figsize=(10, 5))

plt.plot(
    x_values,
    y_values,
    color="#2563EB",
    linewidth=2.5,
    label="p(λ)"
)

plt.axhline(
    0,
    color="black",
    linewidth=1
)

# Numerical eigenvalues
eigenvalues = [complex(v) for v in A.eigenvals().keys()]

real_eigenvalues = [
    value.real for value in eigenvalues
    if abs(value.imag) < 1e-10
]

if real_eigenvalues:
    plt.scatter(
        real_eigenvalues,
        [0] * len(real_eigenvalues),
        color="#DC2626",
        s=80,
        zorder=5,
        label="Real eigenvalues"
    )

plt.title("Characteristic Polynomial p(λ)")
plt.xlabel("λ")
plt.ylabel("p(λ)")
plt.grid(alpha=0.25)
plt.legend()
plt.show()


# ------------------------------------------------
# 13. VISUALIZATION — TIMING COMPARISON
# ------------------------------------------------
methods = ["Direct\nNumPy", "C-H\nReduction"]
times = [direct_time, ch_time]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    methods,
    times,
    color=["#60A5FA", "#34D399"],
    width=0.55
)

plt.ylabel("Average Time (seconds)")
plt.title("Computational Time: Direct vs Cayley-Hamilton")
plt.grid(axis="y", alpha=0.25)

for bar, value in zip(bars, times):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.2e}",
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.show()


# ------------------------------------------------
# 14. BENCHMARK TEST DATASET
# ------------------------------------------------
print("\n" + "=" * 70)
print("8. BENCHMARK DATASET")
print("=" * 70)

print("\nThe following matrices can be used as test cases:")

for i, matrix in enumerate(BENCHMARK_DATA, start=1):
    print(f"\nTest Case {i}:")
    print(np.array(matrix))

print("\nBenchmark verification:")

benchmark_results = []

for i, matrix in enumerate(BENCHMARK_DATA, start=1):

    B = sp.Matrix(matrix)
    B_poly = sp.expand(B.charpoly(x).as_expr())

    # Evaluate characteristic polynomial at B
    B_poly_obj = sp.Poly(B_poly, x)
    B_coeffs = B_poly_obj.all_coeffs()

    B_CH = sp.zeros(B.rows)

    for j, coefficient in enumerate(B_coeffs):
        degree = B.rows - j
        B_CH += coefficient * (B ** degree)

    B_CH = sp.simplify(B_CH)

    # A^8 through C-H
    B8_CH, _ = matrix_power_ch(B, 8, B_poly)

    # Direct A^8
    B8_direct = B ** 8

    ch_pass = B_CH == sp.zeros(B.rows)
    power_pass = sp.simplify(B8_CH - B8_direct) == sp.zeros(B.rows)

    benchmark_results.append(
        [i, f"{B.rows}x{B.cols}", ch_pass, power_pass]
    )

print("\nTest Case | Size | p(A)=0 | A^8 Check")
print("-" * 45)

for row in benchmark_results:
    print(
        f"{row[0]:^9} | "
        f"{row[1]:^4} | "
        f"{'PASS':^7} | "
        f"{'PASS':^10}"
    )


# ------------------------------------------------
# 15. FINAL MATHEMATICAL CONCLUSION
# ------------------------------------------------
print("\n" + "=" * 70)
print("FINAL CONCLUSION")
print("=" * 70)

print("""
The Cayley-Hamilton theorem was applied to the input matrix.

1. The characteristic polynomial p(λ) = det(λI - A) was calculated.
2. λ was replaced by A to obtain p(A).
3. p(A) = 0 was verified.
4. When det(A) ≠ 0, A^(-1) was derived from the
   Cayley-Hamilton equation.
5. A^8 was reduced using the characteristic polynomial.
6. The Cayley-Hamilton result for A^8 was compared with
   direct matrix-power computation.
7. Computational timings were measured.
8. A benchmark dataset was included for testing.

This demonstrates how the Cayley-Hamilton theorem can reduce
high powers and matrix functions to combinations of lower powers.
""")

print("=" * 70)
print("                 C7 PROJECT COMPLETE")
print("=" * 70)
```
